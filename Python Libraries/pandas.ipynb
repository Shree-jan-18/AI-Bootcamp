PANDAS ASSIGNMENT

```python
import pandas as pd
```

Series


```python
marks = pd.Series([78, 85, 90, 67, 88])
print(marks)
```

Dataframe


```python
data = {
    "Name": ["Ram", "Sita", "Hari"],
    "Age": [20, 21, 19],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)
print(df)
```

Reading a CSV(Comma Seperated Values) File

```python
df = pd.read_csv("student.csv")
print(df)
```

Dataframe presentation

```python
print("\nFirst 5 rows of the DataFrame:")
print(df.head())
print("\n\nLast 5 rows of the DataFrame:")    
print(df.tail())
print("\n\nRandom sample of the DataFrame:")
print(df.sample())
print("\n\nDataFrame information:")
df.info()  
print("\n\nStatistical summary of the DataFrame:")
df.describe()
```

```python
df.shape
print(df.columns)
df.dtypes
```

Column Selection

```python
print(df[["Name", "Math"]])
```

```python
print(df[(df["Department"] == "IT") & (df["Science"] > 85)])
```

Null Value Check

```python
print(df.isnull().sum())
print("\n")
print(df.fillna(0))
print("\n")
df.dropna()
```

Adding New Columns

```python
df["Total"] = df["Math"] + df["Science"] + df["English"]

df["Percentage"] = df["Total"] / 3

print(df)
```
