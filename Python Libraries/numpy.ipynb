#NUMPY ASSIGNMENT

```python
import numpy as np
```

Methods for creating arrays


```python
arr1 = np.array([1,2,3,4,5,6])
print(arr1)
```

```python
arr2 = np.zeros((2,3))
print(arr2)
```

```python
arr3 = np.ones((2,4))
print(arr3)
```

```python
arr4 = np.random.randint(1,100, size=(2,3))
print(arr4)
```

Array Properties

```python
vtr = np.array([1, 2, 3, 4])
mtr = np.array([[1, 2, 3], [4, 5, 6]])

print("Vector:", vtr)
print("Matrix:\n", mtr)
print("\n")
print("Shape of vector:", vtr.shape)
print("Shape of matrix:", mtr.shape)
print("Dimensions(axes) of matrix:", mtr.ndim)
print("Data type:", mtr.dtype)
print("Total elements:", mtr.size)
```

Indexing of Array


```python
arr1 = np.array([10,20,30,40,50])

print(arr1[0])
print(arr1[2])
print(arr1[-1])
```

```python
arr2 = np.array([[1,2,3],
                [4,5,6]])

print(arr2[1,2])
print(arr2[0,2])
```

Slicing


```python
arr = np.array([10,20,30,40,50,60])

print(arr[1:4])
print(arr[:3])
print(arr[3:])
print(arr[::-1])
```

Arithmetic operations

```python
a = np.array([10,20,30])
b = np.array([1,2,3])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**2)
```

Statistical Operations

```python
marks = np.array([70,80,90,60,85])

print(np.sum(marks))
print(np.mean(marks))
print(np.median(marks))
print(np.std(marks))
print(np.var(marks))
print(np.min(marks))
print(np.max(marks))
```

Example

```python
marks = np.array([78,65,92,85,74])

print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
print("Above 75:", marks[marks>75])
```
