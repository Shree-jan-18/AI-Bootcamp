#SEABORN ASSIGNMENT

```python
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("student.csv")
```

THEME SETTING

```python
sns.set_theme(style="darkgrid")
```

LINE PLOT

```python
sns.lineplot(
    data=df,
    x="ID",
    y="Math"
)

plt.show()
```

BAR PLOT

```python
sns.barplot(
    data=df,
    x="Department",
    y="Math"
)

plt.show()
```

COUNT PLOT

```python
sns.countplot(
    data=df,
    x="Department"
)

plt.show()
```

SCATTER PLOT

```python
sns.scatterplot(
    data=df,
    x="Math",
    y="Science",
    hue="Department"
)

plt.show()
```

HISTOGRAM PLOT

```python
sns.histplot(
    data=df,
    x="Math",
    bins=6
)

plt.show()
```

BOX PLOT

```python
sns.boxplot(
    data=df,
    x="Department",
    y="Science"
)

plt.show()
```

HEAT MAP

```python
corr = df.select_dtypes(include="number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.show()
```

REGRESSION PLOT

```python
sns.regplot(
    data=df,
    x="Math",
    y="Science"
)

plt.show()
```
