#MATPLOT

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df  = pd.read_csv("student.csv")
df.fillna(0)
```

LINE PLOT

```python
plt.figure(figsize=(5,3))

plt.plot(df["ID"], df["Math"], marker='o')

plt.title("Math Marks of Students")
plt.xlabel("Student ID")
plt.ylabel("Math Marks")

plt.grid(True)

plt.show()
```

BAR PLOT

```python
plt.figure(figsize=(8,3))

plt.bar(df["Name"], df["Science"])

plt.title("Science Marks")

plt.xlabel("Students")
plt.ylabel("Marks")

plt.xticks(rotation=45)

plt.show()
```

SCATTER PLOT

```python
plt.figure(figsize=(5,3))

plt.scatter(df["Math"], df["Science"])

plt.title("Math vs Science")

plt.xlabel("Math")
plt.ylabel("Science")

plt.show()
```

HISTOGRAM

```python
plt.figure(figsize=(5,3))

plt.hist(df["Math"], bins=6)

plt.title("Distribution of Math Marks")

plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

PIE CHART

```python
department_count = df["Department"].value_counts()

plt.figure(figsize=(5,5))

plt.pie(
    department_count,
    labels=department_count.index,
    autopct="%1.1f%%"
)

plt.title("Students by Department")

plt.show()
```

BOX PLOT

```python
plt.figure(figsize=(5,4))

plt.boxplot(df["Attendance"].dropna())

plt.title("Attendance")

plt.show()
```
